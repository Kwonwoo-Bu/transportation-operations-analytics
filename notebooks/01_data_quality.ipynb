```python
# Check how many flights were neither cancelled nor diverted.
# These flights are the main candidates for the operational delay analysis.

df[(df["Cancelled"] == 0) & (df["Diverted"] == 0)].shape
```

```python
# Keep only flights that were not cancelled and were not diverted.
# These flights have a normal departure and arrival and are more appropriate
# for the main delay analysis.


df_analysis = df[(df["Cancelled"] == 0) & (df["Diverted"] == 0)].copy()
```

```python
# Check the number of rows after removing cancelled and diverted flights.

df_analysis.shape
```

```python
# Review the available columns in the filtered dataset.

df_analysis.columns.tolist()
```

```python
# Select the variables needed for the final cleaned dataset and analysis.

selected_columns = [
    "FlightDate",
    "Year",
    "Month",
    "DayOfWeek",
    "Reporting_Airline",
    "Origin",
    "OriginCityName",
    "OriginState",
    "Dest",
    "DestCityName",
    "DestState",
    "CRSDepTime",
    "DepTime",
    "DepDelay",
    "DepDelayMinutes",
    "DepDel15",
    "DepTimeBlk",
    "TaxiOut",
    "CRSArrTime",
    "ArrTime",
    "ArrDelay",
    "ArrDelayMinutes",
    "ArrDel15",
    "ArrTimeBlk",
    "TaxiIn",
    "AirTime",
    "Distance",
    "CarrierDelay",
    "WeatherDelay",
    "NASDelay",
    "SecurityDelay",
    "LateAircraftDelay"
]

df_clean = df_analysis[selected_columns].copy()
```

```python
# Check the shape of the cleaned dataset.

df_clean.shape
```

```python
# Check the data types before making any transformations.

df_clean.dtypes
```

```python
# Convert FlightDate to datetime so it can be used consistently
# for date-based analysis.

df_clean["FlightDate"] = pd.to_datetime(df_clean["FlightDate"])

df_clean["FlightDate"].dtype
```

```python
# Review the distribution of scheduled and actual departure and arrival times.

df_clean[["CRSDepTime", "DepTime", "CRSArrTime", "ArrTime"]].describe()
```

```python
# Time Feature
# `DepTimeBlk` is provided by BTS as an hourly departure-time block. 
# We also created `DepHour` from `CRSDepTime` to make hour-level analysis easier.

df_clean["DepHour"] = df_clean["CRSDepTime"] // 100

df_clean["DepHour"].head()
```

```python
# Check the number of flights in each scheduled departure hour.

df_clean["DepHour"].value_counts().sort_index()
```

```python
# Create a route identifier using the origin and destination airport codes.
# This will be used for route-level analysis.

df_clean["Route"] = df_clean["Origin"] + "-" + df_clean["Dest"]

df_clean["Route"].nunique()
```

```python
# Check the minimum and maximum values of the new departure-hour feature.

df_clean["DepHour"].min(), df_clean["DepHour"].max()
```

```python
# Check whether any route values are missing.

df_clean["Route"].isna().sum()
```

```python
# Review the new time and route features.

df_clean[["FlightDate", "CRSDepTime", "DepHour", "Origin", "Dest", "Route"]].head(10)
```

```python
from pathlib import Path

# Find all monthly raw CSV files so the complete 2025 dataset can be combined.

raw_files = sorted(Path("../data/raw").glob("*.csv"))

print("number of csv files", len(raw_files))

for file in raw_files:
    print(file.name)
```

```python
# Read the required columns from each monthly file.
# Only the selected variables are loaded to reduce unnecessary memory usage.

monthly_data = []

for file in raw_files:
    print("Reading:", file.name)

    monthly_df = pd.read_csv(
        file,
        usecols=selected_columns + ["Cancelled", "Diverted"],
        low_memory=False
    )

    monthly_data.append(monthly_df)

    print("Rows:", len(monthly_df))
```

```python
# Combine all monthly files into a single 2025 dataset

df_2025= pd.concat(monthly_data, ignore_index=True)

print(df_2025.shape)
```

```python
# Remove cancelled and diverted flights from the complete 2025 dataset
# so the final analysis focuses on flights with normal operations.

df_analysis = df_2025[(df_2025["Cancelled"] == 0) & (df_2025["Diverted"] == 0)].copy()

print(df_analysis.shape)
```

```python
# Create the final cleaned dataset using the selected analysis columns.

df_clean = df_analysis[selected_columns].copy()

df_clean.shape
```

```python
# Check the data types of the final cleaned dataset.

df_clean.dtypes
```

```python
# Convert FlightDate to datetime in the complete 2025 dataset.

df_clean["FlightDate"] = pd.to_datetime(df_clean["FlightDate"])

df_clean["FlightDate"].dtype
```

```python
# Review the distribution of scheduled and actual departure and arrival times
# after combining the full 2025 dataset.

df_clean[["CRSDepTime", "DepTime", "CRSArrTime", "ArrTime"]].describe()
```

```python
# Time Feature
# DepTimeBlk is provided by BTS as an hourly departure-time block.
# We also create DepHour from CRSDepTime to make hour-level analysis easier.

df_clean["DepHour"] = df_clean["CRSDepTime"] // 100

df_clean["DepHour"].head()
```

```python
# Check the distribution of flights across departure hours.

df_clean["DepHour"].value_counts().sort_index()
```

```python
# Create the route identifier for the complete 2025 dataset.

df_clean["Route"] = df_clean["Origin"] + "-" + df_clean["Dest"]

df_clean["Route"].nunique()
```

```python
# Check the range of the departure-hour feature.

df_clean["DepHour"].min(), df_clean["DepHour"].max()
```

```python
# Confirm that the route feature does not contain missing values.

df_clean["Route"].isna().sum()
```

```python
# Review the final time and route features.

df_clean[["FlightDate", "CRSDepTime", "DepHour", "Origin", "Dest", "Route"]].head(10)
```

```python
# Save the final cleaned dataset as a CSV file for use in later analysis
# and other tools.

df_clean.to_csv(Path("../data/processed")/"transportation_2025_clean.csv", index=False)
```
